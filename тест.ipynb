```python
df = pd.read_csv('данные.csv', sep=None, engine='python')
```

```python
first_orders = df.groupby('user_id')['date'].min().reset_index()
first_orders.columns = ['user_id', 'first_visit']
```

```python
cohort_nov1 = first_orders[first_orders['first_visit'] == '2023-11-01']
cohort_users = cohort_nov1['user_id'].unique()
initial_count = len(cohort_users)
```

```python
returned_day1 = df[(df['user_id'].isin(cohort_users)) & (df['date'] == '2023-11-02')]
returned_count = returned_day1['user_id'].nunique()
```

```python
retention_d1 = (returned_count / initial_count) * 100

print(f"Пользователей пришло 1 ноября: {initial_count}")
print(f"Вернулось 2 ноября: {returned_count}")
print(f"Retention 1-го дня: {retention_d1:.2f}%")
```

```python
import pandas as pd

file_path = 'тест.xlsx'

df_audience = pd.read_excel(file_path, sheet_name=0)

df_ab = pd.read_excel(file_path, sheet_name=1)

df_third = pd.read_excel(file_path, sheet_name=2)

print("Колонки во второй вкладке:", df_ab.columns.tolist())
```

```python
import pandas as pd
from scipy import stats

for exp_num in df_ab['experiment_num'].unique():
    current_exp = df_ab[df_ab['experiment_num'] == exp_num]
    
    control = current_exp[current_exp['experiment_group'] == 'control']['revenue']
    test = current_exp[current_exp['experiment_group'] == 'test']['revenue']
    
    arpu_control = control.mean()
    arpu_test = test.mean()
    
    t_stat, p_val = stats.ttest_ind(control, test, equal_var=False)
    
    print(f"--- Эксперимент №{exp_num} ---")
    print(f"ARPU Контроль: {arpu_control:.2f}")
    print(f"ARPU Тест: {arpu_test:.2f}")
    print(f"p-value: {p_val:.4f}")
    
    if p_val < 0.05:
        print("Результат: СТАТИСТИЧЕСКИ ЗНАЧИМО. Разница есть.")
    else:
        print("Результат: Незначимо. Разница случайна.")
    print("\n")
```
